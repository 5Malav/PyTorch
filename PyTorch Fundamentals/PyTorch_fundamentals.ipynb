```python
# Two main ways of perfoming multiplication in
# Neural Network and Deep Learning

# 1.) Element vise Multiplication

# 2.) Matrix Multiplication - Dot Product

# There are two main rules that perfoming matrix
# Multiplication needs to satisfy:

# 1.) The **inner dimension must match

# (3,2) * (3,2) will not work
# (2,3) * (3,2) will work
# (3,2) * (2,3) will work

# 2.) The resulting matrix has the shape of the
# ** Outer dimension

# (2,3) * (3,2) = (2,2)
# (3,2) * (2,3) = (3,3)


```

# PyTorch Reproducability (Taking the Random out of Random)

```python
# Ensuring Reproducability in PyTorch means making sure
# our results stay the same every time we run the code,
# even if it uses randomness
# e.g:- weight initialization, shuffling data, dropout
```

```python
# In short how Neural Network learns :-

# start with random numbers -> tensor opersations->
# update random numbers to try and make them better
# representations of the data -> repeat the process

# To reduce the randomness in neural network and
# PyTorch comes the concept of a "random seed"


```

```python
# create a tensor

import torch

random_tensor_A =  torch.rand(3,4)
random_tensor_B =  torch.rand(3,4)

print(random_tensor_A)
print("\n")
print(random_tensor_B)
print("\n")
print(random_tensor_A==random_tensor_B)
```

```python
# let's create random but reproducible tensors

import torch
import random
import numpy as np

# set the random seed
seed=18

torch.manual_seed(seed)

random.seed(seed)
np.random.seed(seed)

random_tensor_C =  torch.rand(3,4)
random_tensor_D =  torch.rand(3,4)

print(random_tensor_C)
print("\n")
print(random_tensor_D)
print("\n")
print(random_tensor_C==random_tensor_D)
```

```python
# setup seed each time before creating a tensor

torch.manual_seed(seed)
random_tensor_C =  torch.rand(3,4)

torch.manual_seed(seed)
random_tensor_D =  torch.rand(3,4)

print(random_tensor_C)
print("\n")
print(random_tensor_D)
print("\n")
print(random_tensor_C==random_tensor_D)
```
